## Math with Integers and Floats

### Understanding the difference between `/` and `//` (and how this impacts datatypes)

Write an expression to illustrate the answer to each question below. Make sure to run the cell to check your results.

```python
# What is the result of 2 times 2?

print(2 * 2)

# The result of 2 times 2 is 4.
```

```python
# What is the result of 4 divided by 2, using regular division?

print(4 / 2)
print(type(4 / 2))

# The result is 2.0.
# Regular division using / returns a float.
```

```python
# What is the result of 4 divided by 2, using floor division?

print(4 // 2)
print(type(4 // 2))

# The result is 2.
# Floor division using // returns an integer when both numbers are integers.
```

```python
# Ok, how about 2 times 2.5?

print(2 * 2.5)
print(type(2 * 2.5))

# The result is 5.0.
# Because 2.5 is a float, Python returns the answer as a float.
```

```python
# What's 5 divided by 2 using regular division?

print(5 / 2)
print(type(5 / 2))

# The result is 2.5.
# Regular division using / returns a float.
```

```python
# What's 5 divided by 2 using floor division?

print(5 // 2)
print(type(5 // 2))

# The result is 2.
# Floor division using // removes the decimal part and keeps the whole number.
```

```python
# When might you want to use floor division instead of normal division? (Write your answer as a code comment below.)

print(17 // 5)

# I might use floor division when I only want the whole number answer
# and I do not need the decimal part.
# For example, if I have 17 items and each box holds 5 items,
# floor division tells me I can fill 3 complete boxes.
```
