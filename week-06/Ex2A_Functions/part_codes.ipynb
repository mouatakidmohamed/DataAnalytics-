# Exercise 2.A Using built-in functions
## Lab 2. Parsing strings – part codes

```python
def get_supplier_code(part_code):
    # I return everything before the colon because that part is the supplier code.
    return part_code[:part_code.index(':')]
```

```python
def get_product_num(part_code):
    # I return the part between the colon and dash because that is the product number.
    start = part_code.index(':') + 1
    end = part_code.index('-')
    return part_code[start:end]
```

```python
def get_size(part_code):
    # I return everything after the dash because that part is the size.
    return part_code[part_code.index('-') + 1:]
```

```python
part_1 = 'ACME:123-L'
part_2 = 'DI:12345-M'
part_3 = 'ACE:1-12'
```

```python
for code in [part_1, part_2, part_3]:
    supplier = get_supplier_code(code)
    product_num = get_product_num(code)
    size = get_size(code)
    print(f"For {code}, the supplier is {supplier}. Part number {product_num} is available in size {size}.")
# I used a loop here so I do not have to repeat the same print statement three times.
```
