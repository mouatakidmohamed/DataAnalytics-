## Lab 1. Working with names – parsing strings

```python
test_name = 'John Doe'
# I created this test name so I can practice splitting a full name into first and last name.
```

```python
def parse_and_display(name):
    # I am looping through the name to find the space between first name and last name.
    for i in range(len(name)):
        if name[i] == ' ':
            first = name[:i]
            last = name[i+1:]
            print(f"Full name: {name}")
            print(f"First name: {first}")
            print(f"Last name: {last}")
```

```python
parse_and_display(test_name)
# This is where I call the function. The function does not show output until I call it.
```

```python
test_list = ['Grace Flores', 'JB Oinonen', 'Ken Penn']
# I put the names in a list so I can test the same function more than one time.
```

```python
parse_and_display(test_list[0]) # output for Grace Flores
```

```python
parse_and_display(test_list[1]) # output for JB Oinonen
```

```python
parse_and_display(test_list[2]) # output for Ken Penn
```

```python
def parse_and_display2(name):
    # This version works for names with one, two, or three words.
    parts = name.split()
    print(f"Full name: {name}")
    if len(parts) == 1:
        print(f"First name: {parts[0]}")
        print("Last name: None")
    elif len(parts) == 2:
        print(f"First name: {parts[0]}")
        print(f"Last name: {parts[1]}")
    elif len(parts) == 3:
        print(f"First name: {parts[0]}")
        print(f"Middle name: {parts[1]}")
        print(f"Last name: {parts[2]}")
    else:
        print("This name has more than three parts, so I need more logic.")

parse_and_display2('Lorde')
parse_and_display2('Billie Eilish')
parse_and_display2('Megan Thee Stallion')
```
