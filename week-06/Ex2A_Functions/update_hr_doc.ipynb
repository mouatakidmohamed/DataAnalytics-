## Lab 3. Employee data
This lab exercise makes updates to information extracted from an employee database.

```python
hr_list = [
    ('0123', 'HR', 'Rebecca Yang', '69000'),
    ('0121', 'IT', 'Mark Blick', '67000'),
    ('0068', 'IT', 'Kahna Larsen', '112000'),
    ('0234', 'OPS', 'Jim Smith', '54000')
]
# I created the HR list as tuples inside a list, like the exercise asked.
```

```python
hr_list[1] = ('0121', 'IT', 'Mark Blick-Hawley', '67000')
# Mark changed his last name, so I replaced his tuple with the updated name.
```

```python
print(hr_list)
```

```python
hr_list[3] = ('0234', 'CS', 'Jim Smith', '60000')
# Jim moved departments, so I updated the department and salary at the same time.
```

```python
print(hr_list)
```

```python
for emp_num, dept, name, salary in hr_list:
    print(f"{emp_num} | {dept} | {name} | {int(salary):,}")
# I converted salary to int so the comma formatting works correctly.
```
