Working with datetime functions.

```python
import datetime
```

# Today

```python
today = datetime.datetime.now()
# I used now() because I want today's current date and time.
```

```python
print(today.strftime("Today is %A, %B %d, %Y"))
# strftime lets me control how the date displays.
```

# My birthday

```python
my_birthday = datetime.date(1998, 1, 1)
# I used a sample birthday here. I can change it to my real birthday if needed.
```

```python
print(my_birthday.strftime("My birthday is %m/%d/%Y"))
```

# Next 90 Days

```python
ninety_d = today + datetime.timedelta(days=90)
```

```python
print(ninety_d.strftime("90 days from today is %B %d, %Y"))
```

# Time for Dinner

```python
dinner_time = datetime.time(7, 30)
```

```python
print(dinner_time.strftime("Let’s meet for dinner at %I:%M %p"))
```
