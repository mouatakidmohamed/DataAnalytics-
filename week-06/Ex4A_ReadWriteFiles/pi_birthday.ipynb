```python
pi = open('pi_million_digits_bonus.txt', 'r')
pi_lines = pi.readlines()
pi.close()
# I save the lines in a list so I can search through them.
```

```python
def pi_bday_check(bday):
    for i in pi_lines:
        if bday in i:
            print('Your birthday is in the first million digits of pi')
            break
```

```python
def pi_bday_check2(bday):
    birthday_found = 0
    for i in pi_lines:
        if bday in i:
            print('Your birthday is in the first million digits of pi')
            birthday_found += 1
            break
    if birthday_found != 1:
        print('Sorry, your birthday was not found in the first million digits of pi')
```

```python
pi_string = ''
for i in pi_lines:
    pi_string += i.strip()
print(pi_string[0:300])
```

```python
def pi_bday_check3(bday):
    birthday_found = 0
    bday_position = ''
    for i in pi_lines:
        if bday in i:
            print('Your birthday is in the first million digits of pi')
            birthday_found += 1
            break
    if birthday_found != 1:
        print('Sorry, your birthday was not found in the first million digits of pi')
    else:
        bday_position = pi_string.index(bday) - 1
        # I subtract 1 because indexing starts at zero and pi starts with 3 plus a decimal point.
        print(f'Your birthday begins at decimal place {bday_position}')
```
