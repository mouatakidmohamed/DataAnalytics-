# Exercise 5.A NumPy Fundamentals
This notebook practices NumPy arrays, dimensions, data types, transposing, slicing, and creating arrays of ones.

```python
import numpy as np
```

## Array Basics

```python
fruits = ['apple', 'banana', 'cherry', 'date']
# I started with a normal Python list before turning it into a NumPy array.
```

```python
arr_1 = np.array(fruits)
print(arr_1)
```

```python
print(arr_1.dtype)
# My result is a Unicode string type like <U6.
# U means Unicode string, and 6 means the longest string in the array has 6 characters.
```

```python
arr_2 = np.array([[10, 11, 12], [13, 14, 15]])
print(arr_2)
```

```python
print(arr_2.T)
# .T prints a transposed view of arr_2.
```

```python
print(arr_2)
# No, arr_2 is not actually changed. The stored array is still the original shape.
```

```python
arr_3 = np.array([[1, 3, 5], [2, 4, 6]])
print(arr_3)
```

```python
arr_4 = np.array([arr_3[0, 2], arr_3[1, 1]])
print(arr_4)
# I used indexing to pull 5 and 4 from arr_3.
```

```python
arr_5 = np.ones((2, 2, 2))
print(arr_5)
# This creates a 3D array of all 1s with two values in each dimension.
```
