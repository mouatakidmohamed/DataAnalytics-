This lab demonstrates how to create payroll tax functions and test them with different employees.

```python
def get_soc_sec_tax(gross_pay):
    # Social Security tax is 6.2% of gross pay.
    return gross_pay * 0.062
```

```python
def get_medicare_tax(gross_pay):
    # Medicare tax is 1.45% of gross pay.
    return gross_pay * 0.0145
```

```python
def get_federal_tax(gross_pay, withholding_code):
    # I use if/elif/else because the tax rate changes based on the withholding code.
    if withholding_code == 0:
        rate = 0.23
    elif withholding_code == 1:
        rate = 0.21
    elif withholding_code == 2:
        rate = 0.195
    elif withholding_code == 3:
        rate = 0.185
    else:
        rate = 0.18 - ((withholding_code - 4) * 0.005)
    return gross_pay * rate
```

```python
gross_pay = 750
code = 0
print(f"Person1 Social Security: ${get_soc_sec_tax(gross_pay):.2f}")
print(f"Person1 Medicare: ${get_medicare_tax(gross_pay):.2f}")
print(f"Person1 Federal Tax: ${get_federal_tax(gross_pay, code):.2f}")
```
