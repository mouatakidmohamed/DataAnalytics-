Working with defining custom functions.

```python
def favorite_things(name, movie, music):
    # This function returns my favorite things in three lines.
    return f"Name: {name}
Favorite movie: {movie}
Favorite music: {music}"

print(favorite_things('Mohamed', 'The Pursuit of Happyness', 'Afrobeat'))
```

```python
def why_im_here(purpose):
    # This function returns my reason for joining the program.
    return purpose

print(why_im_here('I joined Year Up United Data Analyst program to build strong data skills and prepare for a professional internship.'))
```

```python
def favorite_place(place, why):
    # This function explains one favorite place and the reason why.
    return f"One of my favorite places is {place} because {why}."
```

```python
print(favorite_place('the beach', 'it helps me relax and clear my mind'))
```

```python
print(favorite_place('a quiet coffee shop', 'I can focus and study better there'))
```
